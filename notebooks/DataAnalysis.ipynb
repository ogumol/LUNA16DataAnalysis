## Diameter correlation

```python
import pandas as pd
import matplotlib.pyplot as plt

file_path = 'MetadatabyNoduleMaxVoting (1).xlsx'
df = pd.read_excel(file_path)

benign_diam = df[df['Diagnosis'] == 'Benign']['diameter_mm'].dropna()
malign_diam = df[df['Diagnosis'] == 'Malign']['diameter_mm'].dropna()

plt.figure(figsize=(10, 6))

plt.hist(benign_diam, bins=20, alpha=0.5, color='blue', label='Benign', edgecolor='black')
plt.hist(malign_diam, bins=20, alpha=0.5, color='red', label='Malign', edgecolor='black')

plt.title('Nodule by diameter')
plt.xlabel('Diameter Intervals (mm)')
plt.ylabel('Number of Nodules')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

## Subtlety correlation

```python
import pandas as pd
import matplotlib.pyplot as plt

file_path = 'MetadatabyNoduleMaxVoting (1).xlsx'
df = pd.read_excel(file_path)

freq_table = pd.crosstab(df['Subtlety_value'], df['Diagnosis'])

freq_table.plot(kind='bar', figsize=(8, 6), color=['blue', 'red'], alpha=0.6)

plt.title('Nodule by subtlety')
plt.xlabel('Subtlety value')
plt.ylabel('Number of nodules')
plt.xticks(rotation=0)


plt.show()
```
